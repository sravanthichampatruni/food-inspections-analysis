# 03 Load into PostgreSQL (optional helper)
Run `sql/01_load_tables.sql` first to create the tables. The password is typed in when you run the cell and is never saved in the file.

```python
import getpass
import pandas as pd
from sqlalchemy import create_engine

username = "postgres"
password = getpass.getpass("Postgres password: ")
host, port, database = "localhost", 5432, "postgres"
engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")
```

```python
ins = pd.read_csv("../data/processed/inspections_clean.csv", parse_dates=["inspection_date"])
vio = pd.read_csv("../data/processed/violations_clean.csv")
ins.to_sql("inspections", engine, if_exists="append", index=False, chunksize=10000)
vio.to_sql("violations", engine, if_exists="append", index=False, chunksize=10000)
```

```python
print(pd.read_sql("SELECT COUNT(*) FROM inspections", engine))
print(pd.read_sql("SELECT COUNT(*) FROM violations", engine))
```
