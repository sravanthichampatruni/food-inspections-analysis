```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()
ins = pd.read_csv("../data/processed/inspections_clean.csv", parse_dates=["inspection_date"])
vio = pd.read_csv("../data/processed/violations_clean.csv")
ins.shape, vio.shape
```

## 1. Fail rate by facility type (at least 100 inspections)

```python
by_type = (ins.groupby("facility_type")["is_fail"]
           .agg(["count", "mean"]).query("count >= 100")
           .sort_values("mean", ascending=False).head(15))
by_type["mean"].mul(100).plot(kind="barh", figsize=(8, 6), title="Fail rate (%) by facility type")
plt.gca().invert_yaxis()
plt.show()
```

## 2. Fail rate by year

```python
ins["inspection_year"] = ins["inspection_date"].dt.year
(ins.groupby("inspection_year")["is_fail"].mean() * 100).plot(marker="o", title="Fail rate (%) by year")
plt.show()
```

Finding: 

## 3. Most common violations in failed inspections

```python
failed_ids = ins.loc[ins["is_fail"], "inspection_id"]
top = (vio[vio["inspection_id"].isin(failed_ids)]
       .groupby(["violation_code", "violation_desc"]).size()
       .sort_values(ascending=False).head(10))
top
```
